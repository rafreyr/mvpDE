# Section for Full TMDB Movies Dataset 2024 
(RAW >> BRONZE)

```python
# READ TMDB_movie_dataset_v11.csv file from 0_raw.film_data
film_raw = (
    spark.read
    .format("csv") # file is comma separated csv
    .option("header", "true") # 1st line of csv is header
    .option("inferSchema", "false") # not infer column data types (reads as string)
    .option("multiline","true") # single cells can have line breaks (prevents breaking at '\n')
    .option("quote", '"') # defines enclosing quote character
    .option("escape", '"') # ignore double quotes in file columns' text (eg. a description)
    .load("/Volumes/mvpde/0_raw/film_data/TMDB_movie_dataset_v11.csv") # path to csv file
)

# INSERT metadata into films bronze table 
from pyspark.sql.functions import current_timestamp, col
film_bronze = (
    film_raw
    .withColumn("_ingested_at", current_timestamp()) # when the spark pipeline processed the batch
    .withColumn("_source_file_path", col("_metadata.file_path")) # full cloud storage path
    .withColumn("_source_file_name", col("_metadata.file_name")) # short file name
    .withColumn("_source_file_modification_time", col("_metadata.file_modification_time")) # timestamp when the file was dropped into cloud storage
)

# SAVE film_bronze as Delta table
(
    film_bronze
    .write
    .format("delta")
    .mode("append") # sub for ("overwrite") if rerunning notebook
    .saveAsTable("mvpde.1_bronze.film") # path to save
)
```
